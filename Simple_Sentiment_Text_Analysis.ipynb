```python
from textblob import TextBlob
from newspaper import Article

# Step 1: Get and process the article
url = 'https://en.wikipedia.org/wiki/Horse'
article = Article(url)

try:
    article.download()
    article.parse()
    article.nlp()
except Exception as e:
    print("Error downloading or processing the article:", e)
    exit()

# Step 2: Extract and display the summary of the article
text = article.summary
print("=== Article Summary ===")
print(text)

# Step 3: Perform Sentiment Analysis
blob = TextBlob(text)
sentiment = blob.sentiment.polarity  # Sentiment score: -1 to 1
subjectivity = blob.sentiment.subjectivity  # Subjectivity score: 0 to 1

# Step 4: Classify Sentiment
if sentiment > 0.5:
    sentiment_label = "Very Positive"
elif sentiment > 0.1:
    sentiment_label = "Positive"
elif sentiment > 0:
    sentiment_label = "Slightly Positive"
elif sentiment == 0:
    sentiment_label = "Neutral"
elif sentiment > -0.1:
    sentiment_label = "Slightly Negative"
elif sentiment > -0.5:
    sentiment_label = "Negative"
else:
    sentiment_label = "Very Negative"

# Step 5: Display Sentiment Results
print("\n=== Sentiment Analysis ===")
print(f"Sentiment Score: {sentiment:.2f}")
print(f"Subjectivity Score: {subjectivity:.2f}")
print(f"Sentiment Label: {sentiment_label}")

# Count the number of sentences and words in the summary
num_sentences = len(blob.sentences)
num_words = len(blob.words)

# Extract and display keywords
keywords = article.keywords

print("\n=== Additional Analysis ===")
print(f"Number of Sentences in the Summary: {num_sentences}")
print(f"Number of Words in the Summary: {num_words}")
print("Keywords from the Article:")
print(", ".join(keywords))

```
